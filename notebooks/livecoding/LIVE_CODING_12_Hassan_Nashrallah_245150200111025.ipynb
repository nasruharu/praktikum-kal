# **Market Basket Analysis**
Market basket analysis adalah suatu metodologi untuk melakukan analisis buying habit konsumen dengan menemukan asosiasi antar beberapa item yang berbeda, yang diletakkan konsumen dalam shopping basket (keranjang belanja) yang dibeli pada suatu transaksi tertentu. ## **Algoritma Apriori**

Algoritma Apriori adalah suatu algoritma dasar yang diusulkan oleh Agrawal & Srikant pada tahun 1994 untuk penentuan frequent itemsets untuk aturan asosiasi boolean. Algoritma Apriori memberi kita sifat asosiatif dalam transaksi. Ini juga dikenal sebagai Aturan Asosiasi. Aturan asosiasi atau association rule adalah teknik untuk menemukan aturan asosiasi antara suatu kombinasi item. Terdapat 3 metrik untuk mengukur ketepatan aturan, yaitu :


### **1. Support**

## **5. Market Basket Analysis**

Pertama kita mengubah item list menjadi data transaksi dengan menggunakan TransactionEncoder.
Kita akan menggunakan frequent item-sets untuk menghasilkan aturan asosiasi dengan menggunakan algoritma Apriori.


```python
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import association_rules, apriori
```

kita membuat list transaction_list() yang unik sehingga kita dapat mengubah data kita ke dalam format yang benar

```python
transaction_list = []

for i in df['Transaction'].unique():
    tlist = list(set(df[df['Transaction']==i]['Item']))
    if len(tlist)>0:
        transaction_list.append(tlist)
print(len(transaction_list))

transaction_list[:3]
```

kita menggunakan TransactionEncoder untuk merubah bentuk transaction_list()

```python
te = TransactionEncoder()
te_ary = te.fit(transaction_list).transform(transaction_list)
df2 = pd.DataFrame(te_ary, columns=te.columns_)
df2.head()
```

Sekarang kita terapkan Apriori. Kita akan menggunakan parameter min_threshold (nilai ambang batas yang ditentukan) dalam aturan asosiasi untuk metrik lift menjadi 1,0 karena jika kurang dari satu, maka kedua item tersebut kemungkinan tidak akan dibeli bersama. Kita akan mengurutkan nilai berdasarkan keyakinan untuk melihat kemungkinan suatu barang dibeli jika pendahulunya dibeli.

```python
frequent_itemsets = apriori(df2, min_support=0.01, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)
metrix = pd.DataFrame(rules)
metrix[['antecedents','consequents', 'support', 'confidence', 'lift']].sort_values('lift', ascending=False)
```

# **Kesimpulan**

Jelaskan makna tabel di atas berdasar nilai-nilai metrik (support, confidence, dan lift)!

Tabel tersebut adalah hasil association rules dari algoritma Apriori, yang mana menunjukkan pola hubungan antar item. Metrik utama yang dipakai adalah support, confidence, dan lift. Support menunjukkan seberapa sering kombinasi item tersebut muncul dibandingkan seluruh transaksi. Contohnya pada baris pertama, terdapat antecedent Cake dengan consequents Coffee dan Tea, memiliki support sebesar 0.010037. Hal ini berarti sekitar 1% (0.010037 × 100% = 1.0037%) dari seluruh transaksi berisi Cake, Coffee, dan Tea secara bersamaan. Semakin tinggi nilai support, semakin sering kombinasi item itu terjadi dalam dataset.

Confidence menunjukkan peluang item consequents dibeli jika item antecedent sudah dibeli. Contohnya, terdapat baris dengan antecedent Toast dan consequents Coffee, memiliki confidence sebesar 0.704403. Hal ini berarti dari semua transaksi yang membeli Toast, sekitar 70%-nya juga membeli Coffee. Namun, arah aturan juga sangat berpengaruh. Misalnya, dari semua transaksi yang membeli Coffee, hanya sekitar 4.95% yang juga membeli Toast.

Terakhir, Lift menunjukkan seberapa kuat hubungan antar item dibandingkan jika keduanya dibeli secara acak atau tidak berhubungan. Semakin besar nilai Lift, semakin kuat kecenderungan dua item dibeli bersamaan. Contohnya, terdapat antecedent Cake dengan consequents Coffee dan Tea, memiliki lift sebesar 1.937977. Hal ini berarti pembelian Cake memiliki hubungan positif dengan pembelian Coffee dan Tea. Kombinasi ini hampir 1.94 kali lebih mungkin terjadi dibandingkan jika item-item tersebut muncul secara acak.

Berdasarkan tabel tersebut, aturan asosiasi yang terbentuk menunjukkan adanya hubungan positif antar item karena semua nilai lift lebih besar dari 1. Hal ini berarti item-item pada tabel memiliki kecenderungan untuk dibeli bersama. Kombinasi dengan lift tertinggi adalah Cake → Coffee dan Tea, yang menunjukkan bahwa Cake memiliki hubungan cukup kuat dengan pembelian Coffee dan Tea. Sementara itu, aturan Toast → Coffee memiliki confidence tertinggi, yaitu sekitar 70.44%, sehingga dapat dikatakan bahwa pelanggan yang membeli Toast sangat berpotensi juga membeli Coffee. Namun, karena nilai support pada tabel relatif kecil, pola-pola ini tetap perlu dipahami sebagai hubungan yang menarik, tetapi tidak terjadi pada sebagian besar transaksi.
