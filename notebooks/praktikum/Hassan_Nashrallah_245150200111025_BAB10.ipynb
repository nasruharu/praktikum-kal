### Import Library

```python
import pandas as pd
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
import matplotlib.pyplot as plt
```

### Pertanyaan 1


Untuk Pertanyaan 1 pake kode di Prosedur Percobaan

```python
data = {'X': [1, 2, 5, 6, 8],
        'Y': [1, 1, 5, 6, 8]}
df = pd.DataFrame(data)

# single linkage
linked_single = linkage(df, method='single', metric='cityblock')
plt.figure(figsize=(8, 4))
dendrogram(linked_single, labels=df.index, orientation='top')
plt.title('Dendrogram - Single Linkage (Manhattan Distance)')
plt.show()

# complete linkage
linked_complete = linkage(df, method='complete', metric='cityblock')
plt.figure(figsize=(8, 4))
dendrogram(linked_complete, labels=df.index, orientation='top')
plt.title('Dendrogram - Complete Linkage (Manhattan Distance)')
plt.show()
```

Penjelasan

Berdasarkan dendogram tersebut, Single Linkage dan Complete Linkage menghasilkan pola penggabungan cluster yang mirip, namun memiliki tinggi jarak penggabungan dan kedekatan antar cluster yang berbeda. 

Pada Single Linkage, penggabungan cluster dilakukan berdasarkan jarak paling dekat antar anggota dari dua cluster. Terlihat dari dendogram bahwa data 0 dan 1 bergabung terlebih dahulu pada jarak 1, lalu data 2 dan 3 bergabung pada jarak 2. Cluster yang berisi 2 dan 3 kemudian bergabung dengan data 4 pada jarak 4. Terakhir, cluster {0,1} bergabung dengan cluster {2,3,4} pada jarak 7. Dalam hal ini, Single Linkage cenderung lebih cepat menggabungkan cluster selama ada satu pasangan data yang jaraknya dekat.

Sementara itu, pada Complete Linkage, penggabungan cluster dilakukan berdasarkan jarak paling jauh antar anggota dari dua cluster. Hasil penggabungan dan bentuk dendogram pada Complete Linkage mirip dengan Single Linkage. Perbedaannya terlihat pada cluster {2, 3} yang bergabung dengan data 4 pada jarak 6. Lalu, cluster {0,1} bergabung dengan cluster {2,3,4} pada jarak 14. Dibandingkan dengan Single Linkage, jarak penggabungan pada Complete Linkage lebih besar karena metode ini memastikan bahwa seluruh anggota antar-cluster tidak terlalu berjauhan sebelum digabungkan.

### Pertanyaan 2

Untuk Pertanyaan 2 pake Dataset Iris

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
col_names = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'label']
df = pd.read_csv(url, header=None, names=col_names)

X = df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']]

# single linkage
linked_single = linkage(X, method='single', metric='cityblock')
plt.figure(figsize=(16, 8))
dendrogram(linked_single, labels=df.index, orientation='top')
plt.title('Dendrogram - Single Linkage (Manhattan Distance)')
plt.show()

# complete linkage
linked_complete = linkage(X, method='complete', metric='cityblock')
plt.figure(figsize=(16, 8))
dendrogram(linked_complete, labels=df.index, orientation='top')
plt.title('Dendrogram - Complete Linkage (Manhattan Distance)')
plt.show()
```

Penjelasan

Jika digunakan pada dataset Iris, metode Single Linkage dan Double Linkage menghasilkan struktur dendogram yang cukup berbeda. Pada Single Linkage, dendrogram menunjukkan adanya dua kelompok besar yang ditandai dengan warna berbeda, yakni jingga dan hijau. Cluster jingga tampak cukup terpisah dari cluster hijau, tetapi cluster hijau terlihat besar dan memanjang. Hal ini disebabkan oleh Single Linkage yang hanya melihat jarak terdekat antar anggota cluster. Metode ini lebih mudah menggabungkan data, namun kurang membentuk pola pemisahan cluster yang jelas.

Sementara itu, Complete Linkage menghasilkan dendogram dengan pembentukan cluster yang lebih rapi dan terpisah dengan jelas. Dendrogram menunjukkan adanya tiga kelompok besar yang ditandai dengan warna berbeda, yakni jingga, hijau, dan merah. Pola clustering ini sesuai dengan karakteristik dataset Iris yang memang memiliki tiga kelas utama, yakni Iris-setosa, Iris-versicolor, dan Iris-virginica. 

### Pertanyaan 3

Analisis kelebihan dan kekurangan metode Hierarchical Clustering dibandingkan K-Means (Menurut Kalian)

Hierarchical Clustering dan K-Means sama-sama digunakan untuk mengelompokkan data berdasarkan kemiripan, tetapi memiliki cara kerja yang berbeda. Hierarchical Clustering membentuk struktur bertingkat dalam bentuk dendrogram, sedangkan K-Means membagi data langsung ke dalam sejumlah cluster berdasarkan titik pusat atau centroid.

Hierarchical Clustering memiliki kelebihan dalam hasil interpretasi, visualisasi, dan fleksibilitas untuk analisis struktur data. Pada Hierarchical Clustering, jumlah cluster dapat ditentukan setelah melihat dendrogram, berbeda dengan K-Means yang harus menentukan nilai K sejak awal. K-Means dapat menghasilkan hasil berbeda jika titik centroid awal berbeda, sedangkan Hierarchical Clustering umumnya menghasilkan struktur yang konsisten untuk metric dan linkage yang sama. Selain itu, hasil dendogram pada Hierarchical Clustering dapat membantu melihat proses penggabungan data secara bertahap, sehingga hubungan antar-data lebih mudah dianalisis. Tidak seperti K-Means yang hanya menggambarkan hasil clustering data saja dengan menggunakan plot.

Meski begitu, Hierarchical Clustering masih kalah dalam hal efisiensi dan skalabilitas jika dibandingkan dengan K-Means. Hierarchical Clustering membutuhkan perhitungan jarak antar-banyak pasangan data, sehingga secara komputasi lebih berat dibandingkan K-Means. Tidak seperti K-Means, Hierarchical clustering sensitif terhadap outlier dan noise, yang mana dapat memengaruhi struktur dendrogram. Selain itu, jika jumlah data sangat banyak, hasil dendrogram dapat menjadi padat dan sulit dibaca. Hasil diagram plot pada K-Means lebih baik dalam menampilkan hasil clustering dari dataset besar.

### Kesimpulan

Hierarchical clustering adalah metode clustering yang mengelompokkan data dalam bentuk hierarki atau tingkatan, berdasarkan tingkat kemiripan atau jarak antar data. Hierarchical clustering memiliki dua metode dalam membandingkan jarak antar cluster atau linkage, yakni single linkage dan complete linkage. Single linkage menghitung jarak antar cluster berdasarkan pasangan data terdekat. Metode ini akan mencari jarak paling kecil antara dua cluster/data, sehingga membuat data mudah tergabung namun menghasilkan batas cluster yang kurang jelas. Sementara itu, complete linkage menghitung jarak antar cluster berdasarkan pasangan data terjauh. Metode ini akan mencari jarak paling besar antara dua cluster/data, sehingga menghasilkan dendogram yang lebih rapi dengan batas pemisahan cluster yang jelas.
