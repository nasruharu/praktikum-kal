## Tugas A Regresi Linier



### Pertanyaan 1



```python
# import library yang diperlukan
import numpy as np
import matplotlib.pyplot as plt

# pembuatan dataset sintesis
np.random.seed(42)
X = 5 * np.random.rand(200, 1)
y = 7 + 2.5 * X + np.random.randn(200, 1)

# implementasi regresi linier dengan gradient descent
w0 = 0
w1 = 0
alpha = 0.01
epochs = 1000
m = len(X)

for epoch in range(epochs):
    y_pred = w0 + w1 * X
    error = y_pred - y
    grad0 = (2/m) * np.sum(error)
    grad1 = (2/m) * np.sum(error * X)
    w0 -= alpha * grad0
    w1 -= alpha * grad1

# cetak nilai akhir dari w0 dan w1
print(f"w0: {w0}, w1: {w1}")

# hitung dan tampilkan nilai mean squared error (MSE)
y_pred_final = w0 + w1 * X
mse = np.mean((y_pred_final - y) ** 2)
print(f"MSE: {mse}")

# tampilkan plot garis regresi di atas data
plt.scatter(X, y, label="Data")
plt.plot(X, w0 + w1 * X, color='red', label="Regresi Linier")
plt.title("Plot Garis Regresi Linier")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.show()
```

**Penjelasan**

Setelah dibuat dataset sintesis baru berdasarkan rumus dari soal, serta mengimplementasikan regresi linier dengan gradient descent, didapatkan nilai $w_0$ = 7.064 dan $w_1$ = 2.497. Berdasarkan nilai intercept dan slope tersebut, diperoleh rumus persamaan garis regresi $\hat{y} = 7.064 + 2.497x$. Nilai parameter regresi ini hampir mendekati dengan parameter dataset, yang artinya model tersebut berhasil menangkap hubungan linear pada dataset dengan sangat baik.

Didapatkan pula nilai Mean Squared Error (MSE) sebesar 0.934. Nilai tersebut sangat dekat dengan varians noise asli (1.0). Hal ini menandakan model tersebut berhasil mempelajari pola data tanpa terganggu oleh noise yang diberikan. 

Jika dilihat dari grafik plot, garis merah (model) membelah kumpulan data (scatter plot) dengan sangat seimbang di tengah. Tidak ada tanda-tanda garis terlalu ke atas atau terlalu ke bawah. Sebaran titik di sekitar garis merah juga terlihat seragam, yang artinya model memiliki tingkat akurasi yang stabil di semua rentang X.

### Pertanyaan 2

```python
# jika learning rate = 0.001
w0, w1 = 0, 0
alpha = 0.001
loss_history = []

for epoch in range(epochs):
    y_pred = w0 + w1 * X
    error = y_pred - y
    grad0 = (2/m) * np.sum(error)
    grad1 = (2/m) * np.sum(error * X)
    w0 -= alpha * grad0
    w1 -= alpha * grad1

    loss = np.mean(error ** 2)  # rekam MSE tiap epoch
    loss_history.append(loss)
    
y_pred_final = w0 + w1 * X
mse = np.mean((y_pred_final - y) ** 2)
print(f"Learning rate: {alpha}")
print(f"w0: {w0}, w1: {w1}")
print(f"MSE: {mse}")

plt.plot(loss_history, label="alpha = 0.001")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("Konvergensi Gradient Descent")
plt.legend()
plt.show()
```

```python
# jika learning rate = 1.0
w0, w1 = 0, 0
alpha = 1.0
loss_history = []

for epoch in range(epochs):
    y_pred = w0 + w1 * X
    error = y_pred - y
    grad0 = (2/m) * np.sum(error)
    grad1 = (2/m) * np.sum(error * X)
    w0 -= alpha * grad0
    w1 -= alpha * grad1

    loss = np.mean(error ** 2) 
    loss_history.append(loss)
    
y_pred_final = w0 + w1 * X
mse = np.mean((y_pred_final - y) ** 2)
print(f"Learning rate: {alpha}")
print(f"w0: {w0}, w1: {w1}")
print(f"MSE: {mse}")

plt.plot(loss_history, label="alpha = 1.0")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("Konvergensi Gradient Descent")
plt.legend()
plt.show()
```

**Penjelasan**

Jika learning rate diubah menjadi 0.001, terlihat bahwa model mengalami konvergensi yang stabil namun lambat. Model berhasil belajar yang perilaku yang stabil, namun terlalu lambat dengan langkah penurunan yang terlalu kecil. Hal ini ditandai dengan nilai MSE yang masih di angka 4.29, yang menunjukkan bahwa model belum mencapai titik optimal meskipun sudah berjalan 1000 epoch. Sehingga, model butuh lebih banyak epoch untuk benar-benar mempelajari seluruh data.

Sementara itu, jika learning rate diubah menjadi 1.0, terlihat bahwa model menjadi divergen atau gagal dalam belajar. Model mengambil langkah yang terlalu besar, dan setiap langkah baru menghasilkan error yang lebih besar dari sebelumnya. ## Tugas B Regresi Polinomial

### Pertanyaan 1

```python
# pembuatan dataset sintesis
np.random.seed(42)
X = 4 * np.random.rand(150, 1) - 2
y = 5 - 1.2*X + 0.9*(X**2) + np.random.randn(150, 1)

# pembuatan fitur polinomial derajat 2
alpha = 0.01
m = len(X)
X_poly = np.hstack((np.ones((m, 1)), X, X**2))
w_poly = np.zeros((X_poly.shape[1], 1))

# implementasi gradient descent
for epoch in range(epochs):
    y_pred = X_poly @ w_poly
    error = y_pred - y
    gradients = (2/m) * X_poly.T @ error
    w_poly -= alpha * gradients

# tampilkan parameter w0, w1, dan w2
w0, w1, w2 = w_poly.ravel()
print(f"w0: {w0}")
print(f"w1: {w1}") 
print(f"w2: {w2}")
```

**Penjelasan**

Model tersebut adalah implementasi dari regresi polinomial derajat 2 yang menggunakan 150 data acak dari dataset sintesis dengan pola kuadratik $y = 5 + 1.2x + 0.9x^2$. Model tersebut mendapatkan nilai parameter $w_0$ = 5.009, $w_1$ = -1.155, dan $w_2$ = 0.9294, dengan rumus persamaan garis regresi $\hat{y} = 5.009 - 1.155x + 0.9294x^2$. Nilai parameter regresi ini hampir mendekati nilai parameter dataset, yang artinya model tersebut berhasil menangkap hubungan kuadratik pada dataset dengan baik.

### Pertanyaan 2

```python
# model regresi linier
X_lin = np.hstack((np.ones((m, 1)), X))
w_lin = np.zeros((X_lin.shape[1], 1))

for epoch in range(epochs):
    y_pred_lin = X_lin @ w_lin
    error_lin = y_pred_lin - y
    gradients = (2/m) * X_lin.T @ error_lin
    w_lin -= alpha * gradients

# MSE untuk model regresi linier
mse_lin = np.mean((X_lin @ w_lin - y) ** 2)
print(f"MSE linier: {mse_lin}")

# MSE untuk model regresi polinomial
mse_poly = np.mean((X_poly @ w_poly - y) ** 2)
print(f"MSE polinomial: {mse_poly}")
```

**Penjelasan**

Dari hasil MSE di atas, dapat dilihat bahwa model polinomial memiliki MSE yang lebih rendah dibandingkan model linier, yakni berada di angka 1.053. Nilai MSE tersebut menunjukkan bahwa model polinomial hampir sempurna dalam menangkap pola asli di balik noise tersebut. Karena dataset diberikan noise dengan varians 1, maka MSE minimal yang bisa dicapai memang berada di sekitar angka 1. Hal ini juga sejalan dengan hasil parameter polinomial yang sangat mendekati parameter sebenarnya.

Sementara itu, model linier mendapatkan nilai MSE yang cukup tinggi, yakni 2.26. Hal ini terjadi karena model mencoba memaksa pola kuadratik menjadi garis lurus. Model menaikkan nilai $w_0$ menjadi 6.31 untuk mencoba berada di tengah-tengah lengkungan, namun tetap gagal karena ia tidak punya parameter $w_2$ untuk menangkap kelengkungan data. Dalam kasus ini, model linier mengalami underfitting. Model ini terlalu sederhana untuk menangkap data yang kompleks, sehingga prediksi akan meleset jauh dan menghasilkan error yang tinggi.

## Tugas C Eksplorasi dan Analisis

```python
# dataset sintesis
np.random.seed(42)
X = 4 * np.random.rand(150, 1) - 2
y = 5 - 1.2*X + 0.9*(X**2) + np.random.randn(150, 1)

# pembuatan fitur polinomial derajat 5
alpha = 0.001 
epochs = 10000
m = len(X)
X_quintic = np.hstack((np.ones((m, 1)), X, X**2, X**3, X**4, X**5))
w_quintic = np.zeros((X_quintic.shape[1], 1))

# implementasi gradient descent
for epoch in range(epochs):
    y_pred = X_quintic @ w_quintic
    error = y_pred - y
    gradients = (2/m) * X_quintic.T @ error
    w_quintic -= alpha * gradients

# MSE
mse_quintic = np.mean((X_quintic @ w_quintic - y) ** 2)
print(f"MSE polinomial derajat 5: {mse_quintic}")

# pembuatan grafik plot
X_plot = np.linspace(-2, 2, 100).reshape(-1, 1)
X_plot_quintic = np.hstack([X_plot**i for i in range(6)])
y_plot = X_plot_quintic @ w_quintic
plt.scatter(X, y, alpha=0.5, label="Data")
plt.plot(X_plot, y_plot, color='green', label=f"Polinomial Derajat 5")
plt.legend()
plt.show()
```

**Penjelasan**

Jika dilihat dari kurva dan nilai MSE yang dihasilkan, garis kurva dari model dapat mengikuti bentuk pola data dengan baik, tanpa adanya kelokan aneh. Sehingga, model justru fit dengan baik dan tidak overfit, karena model tidak berusaha menyentuh semua titik-titik data, termasuk outlier. Hal ini juga didukung dengan nilai MSE yang rendah, yakni di angka 1.022 dan mendekati standar deviasi noise = 1. 

Secara teori, model ini punya 6 parameter (w0–w5) untuk menjelaskan data yang sebenarnya hanya butuh 3 parameter (w0, w1, w2). Parameter ekstra ini tidak terpakai dan bisa diisi dengan mempelajari noise yang menyebabkan overfit. Namun, dalam kasus ini, ada cukup banyak data yang tersedia sehingga model tidak kehabisan data untuk dipelajari. Selain itu, learning rate yang kecil (0.001) dan epochs yang banyak (10.000) membuat gradient descent punya cukup waktu untuk konvergen perlahan tapi stabil. Hasilnya bobot w3, w4, w5 kemungkinan mendekati nol karena data memang tidak membutuhkan suku derajat tinggi tersebut.

### Pertanyaan 2

```python
def predict(x, w):
    degree = len(w) - 1    
    X_poly = np.hstack([x**i for i in range(degree + 1)])
    y_pred = X_poly @ w

    return y_pred
```

**Penjelasan**

Fungsi predict(x, w) akan menerima data baru dan bobot hasil training sebagai parameter, lalu mengembalikan hasil prediksi. Sebenarnya fungsi ini dapat dibuat dengan sederhana dengan langsung mengembalikan operasi matriks dari $x @ w$. Namun, untuk mempermudah pemanggilan dan pemakaian fungsi, di dalamnya ditambahkan pembuatan fitur polinomial sesuai dengan derajat, yang diambil dari panjang array bobot dikurangi 1. Sehingga, fungsi dapat dipakai untuk semua jenis model polinomial, tanpa perlu ditransformasi. 

### Pertanyaan 3

1. Kapan perlu menggunakan regresi linier?

    Regresi linier digunakan ketika hubungan antara variabel independen (X) dan dependen (y) bersifat linear, ditandai dengan sebaran data yang membentuk pola garis lurus. Metode ini cocok untuk data dengan pola sederhana dan hubungan yang relatif lurus.

2. Kapan regresi polinomial lebih cocok?
    
    Regresi polinomial lebih cocok digunakan ketika data menunjukkan hubungan non-linear, seperti data dengan pola kuadratik, kubik, atau lebih kompleks. Metode ini mampu menangkap kompleksitas data yang tidak dapat direpresentasikan dengan garis lurus.

3. Dampak jumlah epoch dan learning rate terhadap performa

    Jumlah epoch memengaruhi seberapa lama model belajar dari data. Epoch yang terlalu sedikit dapat menyebabkan underfitting, sedangkan epoch yang terlalu banyak dapat menyebabkan overfitting. Sementara itu, learning rate menentukan besar langkah pembaruan parameter. Learning rate yang terlalu kecil dapat membuat pelatihan model menjadi lambat, sedangkan learning rate yang terlalu besar dapat menyebabkan model menjadi tidak stabil, agresif, dan gagal konvergen.
