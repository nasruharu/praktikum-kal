#### Tugas 3: Buatlah program untuk soal nomor 1 dengan rekomendasi Gemini pada Notebook Google Colab. 

```python
import math

def factorize_n(n):
    # Mencari faktor p dan q dari n
    for i in range(2, int(math.sqrt(n)) + 1):
        if n % i == 0:
            return i, n // i
    return None

n = 55
e = 3 # Contoh kunci publik e

# 1. Faktorisasi n
p, q = factorize_n(n)
print(f"Hasil Faktorisasi: p = {p}, q = {q}")

# 2. Hitung nilai Phi (n)
phi_n = (p - 1) * (q - 1)
print(f"Nilai Phi(n): {phi_n}")

# 3. Hitung Kunci Privat (d)
# d adalah modular multiplicative inverse dari e modulo phi_n
def extended_gcd(a, b):
    if a == 0:
        return b, 0, 1
    else:
        g, y, x = extended_gcd(b % a, a)
        return g, x - (b // a) * y, y

def mod_inverse(e, phi):
    g, x, y = extended_gcd(e, phi)
    if g != 1:
        raise Exception('Modular inverse tidak ada')
    else:
        return x % phi

try:
    d = mod_inverse(e, phi_n)
    print(f"Kunci Privat (d) yang berhasil ditemukan: {d}")
except Exception as error:
    print(f"Error: {error}")
```
