### Caesar Cipher

```python
def caesar_enc (plaintext, kunci):
    result = ""
    for char in plaintext:
        if char.isalpha ():
            awal = ord('A') if char.isupper() else ord('a')
            hasil = chr((ord(char) - awal + kunci) % 26 + awal)
            result += hasil
        else:
            result += char
    return result

def caesar_dec (ciphertext, kunci):
    return caesar_enc(ciphertext, -kunci)

plaintext = "KEAMANAN INFORMASI"
kunci = 5

print ("Plaintext:", plaintext)
print ("Kunci:", kunci)

ciphertext = caesar_enc (plaintext, kunci)
print ("Ciphertext:", ciphertext)

hasil_dekripsi = caesar_dec (ciphertext, kunci)
print ("Plaintext:", hasil_dekripsi)
```

### Menentukan Kunci Caesar Cipher

```python
ciphertext = "zlyhunhuYhohzph"

for kunci in range(11):
    plaintext = caesar_dec(ciphertext, kunci)
    print(f"Kunci {kunci}: {plaintext}")
```

```python
def caesar_decrypt(ciphertext, key):
    result = ""
    
    for char in ciphertext:
        if char.isalpha():
            if char.isupper():
                result += chr((ord(char) - ord('A') - key) % 26 + ord('A'))
            else:
                result += chr((ord(char) - ord('a') - key) % 26 + ord('a'))
        else:
            result += char
    
    return result


ciphertext = "zlyhunhuYhohzph"

# coba semua kemungkinan kunci
for key in range(26):
    hasil = caesar_decrypt(ciphertext, key)
    print(f"Kunci {key}: {hasil}")
```

```python
from Crypto.Cipher import AES

key = b'KunciRahasiaSaya'
cipher = AES.new(key, AES.MODE_EAX)

message = b'Pesan rahasia123'
ciphertext = cipher.encrypt(message)
nonce = cipher.nonce

print("Ciphertext:", ciphertext)
print("Nonce:", nonce)
```

```python
cipher2 = AES.new(key, AES.MODE_EAX, nonce=nonce)
plaintext = cipher2.decrypt(ciphertext)

print("Plaintext:", plaintext)
```
