1. Tes CPU — Komputasi Numerik

```python
import numpy as np
import time

# Matrix multiplication besar
start = time.time()
a = np.random.rand(5000, 5000)
b = np.random.rand(5000, 5000)
c = np.dot(a, b)
end = time.time()

print(f"Waktu: {end - start:.2f} detik")
```

2. Tes RAM — Handling Dataset Besar

```python
import pandas as pd
import numpy as np
import time

start = time.time()
df = pd.DataFrame(np.random.rand(5_000_000, 10), columns=[f"col_{i}" for i in range(10)])
result = df.describe()
end = time.time()

print(f"Waktu: {end - start:.2f} detik")
print(f"Ukuran DataFrame: {df.memory_usage().sum() / 1e6:.1f} MB")
```

3. Tes ML Pipeline — Scikit-learn

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
import time

X, y = make_classification(n_samples=100_000, n_features=20)

start = time.time()
clf = RandomForestClassifier(n_estimators=100, n_jobs=-1)
clf.fit(X, y)
end = time.time()

print(f"Waktu training: {end - start:.2f} detik")
```

4. Cek Spesifikasi Sistem (opsional tapi berguna)

```python
import psutil
import platform

print(f"OS: {platform.system()} {platform.release()}")
print(f"CPU: {platform.processor()}")
print(f"Core fisik: {psutil.cpu_count(logical=False)}")
print(f"Core logis: {psutil.cpu_count(logical=True)}")
print(f"RAM total: {psutil.virtual_memory().total / 1e9:.1f} GB")
print(f"RAM tersedia: {psutil.virtual_memory().available / 1e9:.1f} GB")
```

```python
import torch
import time

size = 10000
a = torch.randn(size, size)
b = torch.randn(size, size)

# CPU
start = time.time()
torch.mm(a, b)
print(f"CPU: {time.time() - start:.3f}s")

# GPU
a_gpu = a.cuda()
b_gpu = b.cuda()
torch.cuda.synchronize()
start = time.time()
torch.mm(a_gpu, b_gpu)
torch.cuda.synchronize()
print(f"GPU: {time.time() - start:.3f}s")
```
